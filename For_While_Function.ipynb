# Number guessing game

```python
"""Problem Statement: Create a Python program that implements a simple number
guessing game using a while loop. The program should make use of control
statements such as else, break, and continue."""

import random
 # Generate random number between 1 and 10
secret_number = random.randint(1,10)
attempts = 3
while attempts > 0:
    num = int(input("Guess the number between 1 - 10: "))
    
    if num < 1 or num > 10:
        print("Your guess is out of range. Please guess between 1 and 10.")
        continue
    attempts -= 1

    if num == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    elif num < secret_number:
        print("Too low. Try again.")

    else:
        print("Too high. Try again.")

else:
    print("Better luck next time!")
    print("Secret number is :",secret_number)
```

# Multiplication Table Generator

```python
num=int(input("Enter the number for which you want the multiplication table: "))
for i in range(1,11):
    print(num,"x",i,"=",num*i)
```

# BMI Calculator

```python
def  calculate_bmi(weight, height):
    bmi=weight/(height*height)
    return bmi
    
w=float(input("Enter your weight in kg: "))
h=float(input("Enter your height in meters: "))

result=calculate_bmi(w,h)
print("Your BMI is:", round(result, 2))
```
